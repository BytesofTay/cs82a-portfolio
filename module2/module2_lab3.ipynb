Fahrenheit cutoffs: freezing is 32 or below; cold is above 32 and below 50; mild is 50 to below 65; warm is 65 to below 80; hot is 80 or above.

```python
temps = [61, 65, 72, 58, 70, 74, 68]
boundary_temps = [31, 32, 33, 49, 50, 51, 64, 66, 79, 80, 81]
categories = {}

for temp in temps + boundary_temps:
    if temp <= 32:
        category = "freezing"
    elif temp < 50:
        category = "cold"
    elif temp < 65:
        category = "mild"
    elif temp < 80:
        category = "warm"
    else:
        category = "hot"

    categories[temp] = category
    print(temp, ":", category)
```

```python
expected = {
    31: "freezing", 32: "freezing", 33: "cold",
    49: "cold", 50: "mild", 51: "mild",
    64: "mild", 65: "warm", 66: "warm",
    79: "warm", 80: "hot", 81: "hot"
}

for temp, expected_category in expected.items():
    assert categories[temp] == expected_category

print("All 12 boundary checks passed.")
```
