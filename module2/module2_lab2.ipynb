```python
def multiply_for(a, b):
    total = 0

    for count in range(b):
        total = total + a

    return total

print(multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    total = 0
    counter = 0

    while counter < b:
        total = total + a
        counter = counter + 1

    return total

print(multiply_while(7, 6))
```

```python
tests = [(7, 6, 42), (3, 4, 12), (0, 5, 0), (5, 0, 0), (0, 0, 0), (1, 9, 9)]

for a, b, expected in tests:
    for_result = multiply_for(a, b)
    while_result = multiply_while(a, b)
    print(f"{a} by {b}: {for_result}, {while_result} (expected {expected})")
    assert for_result == expected
    assert while_result == expected
```

Why the while loop stops

The loop stops when the counter reaches b because the condition is no longer true.
