# Module 3 Lab 1: Rescue a Messy Dataset

```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")
print(df.shape)
display(df.head(10))
display(df.dtypes)
```

Three Problems: I initially see that the data types are mis-matched especially on order number 1280 in the zip section. One og the prices , (order_id# 1280), is also missing data. The dates are not formatted properly. 

```python
df.isna().sum()
```

```python
median_price = df["price"].median()
df["price"] = df["price"].fillna(median_price)

print("Fill value:", median_price)
df.isna().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
df["zip"].head(10)
```

```python
dates = pd.to_datetime(df["date"], format="mixed")
df["date"] = dates.astype("datetime64[ns]")

print(df["date"].dtype)
df["date"].head(10)
```

```python
df[df["qty"] < 0]  
```

```python
df["possible_return"] = df["qty"] < 0
display(df[df["possible_return"]])
print(df.shape)
```

```python
df.to_csv("sales_clean.csv", index=False)
print("Saved sales_clean.csv", df.shape)
```
